```python
#Python - 6: Sine and Cosine Function
This notebook shows the functions for sin(x) and cos(x), tabulates their values from 0 to 2π, then prints the first 10 values utilizing  a for loop.
```

```python
import math
```

This cell imports the Math markdown so that sine and cosine are computed. 

```python
def sin_function(x):
"""Returns the sine of x."""
return math.sin(x)
```

The function known as "Sin_function(x):" returns the sine of a given value, in this case being x.

```python
def cos_function(x):
"""Returns the cosine of x."""
return math.cos(x)
```

This function is similar to the sin_function, but instead returns the cosine of a given value, still being x. 

```python
import numpy as np

x_values = np.linspace(0, 2 * math.pi, 1000)
sin_values = [sin_function(x) for x in x_values]
cos_values = [cos_function(x) for x in x_values]
```

This cell creates 1000 values of x between 0 and 2π, thus computing sin(x) and cos(x). 

```python
print("x\t\t sin(x)\t\t cos(x)")
print("---------------------------------------")

for i in range(10):
    x = x_values[i]
    print(f"{x:.6f}\t {sin_function(x):.6f}\t {cos_function(x):.6f}")
```

This code here prints out the 1000 values of x between 0 and 2π, computing sin(x) and cos(x).
