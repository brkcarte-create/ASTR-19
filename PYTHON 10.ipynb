```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
scale = 1.0
random_numbers = np.random.exponential(scale, 1000)
```

```python
plt.figure()

plt.hist(random_numbers, bins=100)

plt.xlabel("Random Number Value")
plt.ylabel("Frequency")
plt.title("Histogram of 1000 Exponential Random Numbers")

plt.savefig("exponential_distribution_histogram.pdf")

plt.show()
```
